### __Plot Coverage Area__

```python
# Initialize variables
min_longitude = -219
max_longitude = -60
min_latitude = 11
max_latitude = 51.5

coverage_map = folium.Map(
    location=[(min_latitude + max_latitude)/2, (min_longitude + max_longitude)/2],
    zoom_start = 2,
    tiles = "CartoDB Positron")

# Add rectangle for the bounding box
folium.Rectangle(
    bounds=[[min_latitude, min_longitude], [max_latitude, max_longitude]],
    color = '#dc7d81',
    weight = 2,
    fill = True,
    fill_opacity = 0.2,
    popup = 'Shifted Bounding Box').add_to(coverage_map)

coverage_map
```
